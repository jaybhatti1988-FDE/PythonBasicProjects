## File Handling Operation

#### 1. Read the content from My Text file in the System.

###### open()--> opening your file
###### First Parameter --> File_name
###### Second Parameter --> Mode
###### Mode --> It decides the way in which we are working with this file
###### Mode --> r --> Reading the content from the File
###### Mode ---> w --> Write the content into the File
###### Mode ---> r+ --> Read the content first and then writing the content into the File
###### Mode ---> w+ ---> Writing the content first and then reading from it '''

```python
# Read your file
# read()---> Reads the content from the File 

first_file=open('example.txt','r')  # read file & store in first_file variable like file
content_file=first_file.read()
print(content_file)
first_file.close()
```

## Writing the Content into the File

##### write()--> it overwrites the content into the text file
##### mode()--->w--> it Content will be overwritten
##### mode()--->a--> it Content will be added from the end

```python
# Write your file
# write()---> write the content from the File 

first_write_file=open('example.txt','w')  # write file & store in first_file variable like file
first_write_file.write('Here So many topics need cover for FDE role.')
first_write_file.close()
```

```python
# Append the file

first_append_write_file=open('example.txt','a')
first_append_write_file.write('\n We are learning about Gen AI pre requisite Python')
first_append_write_file.close()
```

##### Reading the Total Lines , Total Words , Total Characters from a file  ---> readlines()

##### readlines()---> Reads the line in a way where in it is like the lines are stored inside a list ['Hello Good Morning','I am your gen AI Chat box','Let us learn Python First']

```python
def count_file_lines(file):
    with open(file,'r') as files:
        file_lines=files.readlines()
        file_totallines=len(file_lines)
        file_totalwords=sum(len(line.split()) for line in file_lines)
        file_totalcharacters=sum(len(line) for line in file_lines)
        return file_totallines,file_totalwords,file_totalcharacters

file_path='Total_Words.txt'
TotalLines,TotalWords,TotalCharacters=count_file_lines(file_path)    
print(f"Total line of file is:- {TotalLines} \nTotal Words of file is :- {TotalWords}  \nTotal Characters in file is :- {TotalCharacters} ")
```

##### Write the content into a file and then after writing the content into that file we are going to read the content from that file as well -->seek() and w+

##### w+---> First of all as W is thier so it states that firs writing operation will be done and then Reading operation will be done
##### seek()--->It is going to help us out with reading the content after writing from the starting only

```python
with open('Total_Words.txt','w+') as write_and_read_file:
    write_and_read_file.write("\n working in W+ operation in python file handling")
    write_and_read_file.write("\n It's means Reading and Writing file in python file handling")
    write_and_read_file.seek(0)
    content_complete_file_data=write_and_read_file.read()
    print(content_complete_file_data)    
```
