# Pillars of Object Oriented Programming

## Abstraction

#### Abstraction is the concept of Hiding the implementation details and only showing the necessary features of the object .
#### This helps in reducing promgramming complexity and efforts

### How can we achieve abstraction in python

#### 1. Abstract class -->  A class that cannot be instantiated or that means for an abstract class we cant create the object of this class .

#### 2. Abstract method --> Abstract method just because it cant accessed by the object of the abstract class itself thats why we have to call this abstract method in our child class and the implementation of the abstract method is not provided inside the abstract class but must be provided in the child class .

#### The Abstract Method being defined by @abstractmethod

```python
from abc import ABC,abstractmethod

class Vechile(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vechile):

    def start(self):
        print("The car starts with an ignition key")

    def stop(self):
        print("The car stops with break pedals")
        
class ElectricScooter(Vechile):

    def start(self):
        print("Electric Scooter is going to start with a power button")

    def stop(self):
        print("Electric Scooter stops with an electric brake")
```

```python
car=Car()
```

```python
car.start()
```

```python
car.stop()
```

```python
scooter=ElectricScooter()
```

```python
scooter.start()
```

```python
scooter.stop()
```
