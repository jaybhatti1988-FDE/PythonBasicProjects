## Types of Constructors :
#### 1. Default Constructor
#### 2. Parameterized Constructor

##### 1. Default Constructor : It doesnt have the parameters inside it and it is only having the self keyword inside it . It assigns a default value to all of the objects.
##### 2. Parameterized Constructor : It consists of parameters inside it so that every object can have a specific attribute associated to it

```python
class Dog:

    def __init__(self):
        self.name="Golden retriver"
        self.color="light yellow"

    def show_info(self):
        print(f"This dog breed is {self.name} and it color is {self.color}")
```

```python
class Dogs:

    def __init__(self,breedname,color,caegory):
        self.breedname=breedname
        self.color=color
        self.category=caegory

    def showing_info(self):
        print(f"This dog breed is {self.breedname} and it color is {self.color} and it main feature is {self.category}")
```

```python
dogs_1=Dogs("Dober Man","black","alertness")
dogs_2=Dogs("Pug","silver","small,carring & joyful")
```

```python
dogs_1.showing_info()
dogs_2.showing_info()
```

## Types of Methods

#### 1. Instance Methods : Instance Methods operate on Object Attributes .The Instance Method is going to use self keyword to access and modify object attributes .These Methods are called using an object .
#### 2. Static Methods : It is the method of the whole class or the whole blueprint which will remain the same for all of the objects then such kind of Method is being reffered as a Class Method .
##### In Class Method , These methods work with the class level attributes .
##### They are going to use cls instead of self
##### They are called using the class name as well
##### For specific that the current method is a class method we have to use the abbreviation as @classmethod

#### 3. Static Methods : Static Methods do not use self or cls .
##### They work like normal functions but belong to a class .
##### They are going to be designed by a abbreviation @staticmethod
##### static method can also be called by the classname also



```python
# Instance keyword

class Avengers:
    def __init__(self,name,power):
        self.name=name
        self.power=power

    def self_show_details(self):
        print(f"my name is {self.name} and my power is {self.power}. ")

iorn_man=Avengers("Iorn Man","Super Intelligence")
iorn_man.self_show_details()    
```

```python
# Class Method

class Avengers:

    origin="Avengers"

    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def origin_details(cls):
        print(f"my Origin is {cls.origin} ")

    def self_info(self):
        print(f"My name is {self.name} and my power is {self.power}")

iorn_man=Avengers("Iorn Man","Super Intelligence")
iorn_man.self_info()    

iorn_man.power
iorn_man.name
iorn_man.origin_details()
```

```python
Thor=Avengers("Thor","God of Thunder")
Thor.origin_details()
Thor.self_info()
```
