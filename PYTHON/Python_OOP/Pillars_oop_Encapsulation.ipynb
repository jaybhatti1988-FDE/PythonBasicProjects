# Pillars of Object Oriented Programming

## Encapsulation
#### Encapsulation is a way of wrapping data (Variables and Methods) inside a capsule or a class as a Single unit .

#### It restricts the direct access to some of the objects component’s which is a means of preventing accidental inference and misuse of the data .

#### Key concepts of Encapsulation :
#### 1. Public Members
#### 2. Protected Members
#### 3. Private Members
#### getter and setter methods

##### Public Member ---> Accessible from Anywhere or Attributes are accessible outside of the class and these members are accessible by the derived class 
##### public member attribute is directly declared
##### ## dir()---> it takes the input as the object name and gives us this idea that which member is public protected or private.

```python
class IronMan:
    def __init__(self,name,power,weapon):
        self.name=name
        self.power=power
        self.weapon=weapon

    def Avg_Introduction(self):
        print(f"My Name is {self.name} I have weapon is that : {self.weapon}")
```

```python
iron=IronMan("Tony Shark","Super Intelligence","Repulsor Beams")
```

```python
dir(iron)
```

```python
iron.name
```

```python
iron.weapon
```

```python
iron.Avg_Introduction()
```

#### Private Members--> A Private member is not accessible outside the class and not even in the derived class and in order to have the access over these private members or to bring some modification on these members we use getter and setter method .
#### Private Members are declared by using double underscore __ before the name

```python
class Avengers:
    def __init__(self,name,phone_number):
        self.name = name
        self.__phone_number = phone_number

    def intro(self):
        print(f"")
```

```python
avenge=Avengers("captain","091066003")
```

```python
avenge.name
```

```python
avenge._Avengers__phone_number
```

```python
dir(avenge)
```

## Getter Function

## Getter Method ----> it will be helping us out with accessing private data

```python
class Captain:
    def __init__(self,name,address):
        self.name=name
        self.__address=address

## getter Method
    def accessing_add(self):
        print(f"my address is  {self.__address}. ") 
```

```python
captain=Captain("Roger Steves","stepphen wolf street walt")
```

```python
captain.name
```

```python
dir(Captain)
```

```python
captain.accessing_add()
```

#### setter Method---> Setter Method is being used for modifying our private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound

    def sound_info(self):   ## getter method
        print(f"The Animal is going to sound as : {self.__sound}")

    def sound_modify(self,new_sound):
        self.__sound=new_sound
        print(f"New Sound of Animal is going to sound as : {self.__sound}")
```

```python
animal=Animal("Bruno","Wow Wow")
```

```python
animal.name
```

```python
animal.sound_info()
```

```python
animal.sound_modify("Mew Mew")
```

#### Protected Data is the Data which is not accessible outside the class But the protected data is going to be accessible by the derived class or child class
#### For having the protected member you can mention your protected member as _attribute or by using a single underscore

```python
class Grandparents:
    def __init__(self,name,locker_key):
        self.name=name
        self._locker_key=locker_key


# Getter Method
    def locker_info(self):
        print(f"The locker key is that {self._locker_key}")
```

```python
class Parent(Grandparents):
    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)
        

# Getter Method
    def child_reveals(self):
        print(f"The Locker key info is that {self._locker_key}")
```

```python
grandpa=Grandparents("Vinod","90123")
```

```python
grandpa.name
```

```python
dir(grandpa)
```

```python
grandpa._locker_key
```

```python
parent=Parent("Neel","0091024")
```

```python
parent.child_reveals()
```

```python
# Bank System
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return f"Deposited amount is : {amount}. Total balance is : {self.get_balance()}"
        else:
            return "Deposited amount must be positive"

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            return f"Withdrawn amount is : {amount}. Total balance is : {self.get_balance()}"
        else:
            return "Invalid amount or insufficient balance"

acc = BankAccount("Ravi", 1000)
print(acc.deposit(500))
print(acc.withdraw(300))
print(acc.withdraw(5000))
print(acc.deposit(-50))
```
