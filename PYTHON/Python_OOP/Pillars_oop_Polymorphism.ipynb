# Pillars of Object Oriented Programming

## Polymorphism

#### Polymorphism means MANY FORMS in Greek .
#### In OOPs , Polymorphism allows objects of Different classes as if they are the object of the same class & It enables reusability and makes programs more flexible .

#### Types of Polymorphism :

##### Method Overriding (Runtime Polymorphism)
##### Method Overloading (Compile Time Polymorphism)
##### Operator Overloading

#### Method Overriding --> A child class provides a specific implementation of a method that has already been defined into the parent class .This allows Dynamic Selection at Runtime

```python
# Method Overriding
class Grandparents:
    def speak(self):
        print("Hello my name is Jay")

class Parents(Grandparents):
    def speak(self):
        print("Hello my name is Ajay")

class Child(Grandparents):
    def speak(self):
        print("Hello my name is Vijay")
```

```python
def attribute_provider(object):
    return object.speak()
```

```python
grandpa=Grandparents()
```

```python
print(attribute_provider(grandpa))
```

```python
Parents=Parents()
```

```python
attribute_provider(Parents)
```

```python
child=Child()
```

```python
attribute_provider(child)
```

```python
class Shape:
    def area(self):
        return "This method is returns area of the shape"
```

```python
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height

    def area(self):
        return self.width*self.height
```

```python
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
        
    def area(self):
        return 3.14*self.radius*self.radius
```

```python
Shape=Shape()
Shape.area()
```

```python
rectangle=Rectangle(6,4)
rectangle.area()
```

```python
circle=Circle(4)
circle.area()
```
