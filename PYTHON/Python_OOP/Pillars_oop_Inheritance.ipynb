# Pillars of Object Oriented Programming

### Inheritance
#### Inheritance is a Fundamental concept in Object Oriented Programming because of which the child class is able to inherit the properties OR Attributes and functionalities of the parent class.
#### Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparent and definitiely in this hierarchy the grandchild can have some features different from the parent and grandparent as well .

### In Python typically we get to see 3 Types of Inheritance :
#### 1. Single Inheritance
#### 2. Multilevel Inheritance
#### 3. Multiple Inheritance

#### (1) Single Inheritance --> The Features or the Attributes from one parent class will be getting inherited into one child class .
##### Parent Class (Properties--->Attributes and Functionalities)
##### |
##### |
##### Child Class

#### (2) Multilevel Inheritence --> The properties or Methods and Attributes of the grandparent class can be inherited by the parent class and then the properties of the parent class will be inherited by the child class . In a proper heirarchy the Inheritance process is going to work for us .

##### Grandparent (Properties)
##### |
##### |
##### Parent (Properties)
##### |
##### |
##### Child

#### (3) Multiple Inheritance ---> In Multiple Inheritance , The properties of two parent class or multiple parent class will be getting inherited into the child class.

#### Parent_1 (Properties) + Parent_2 (Properties) + Parent_3 (Properties)+....
#### |
#### |
#### Child

```python
# 1. Single Inheritance
class Avengers:
    def __init__(self,name,power):
        self.name =name
        self.power=power
    def self_introduce(self):   
        print(f"My name is {self.name} .")


class Thor(Avengers):
    def __init__(self, name, power,weapon):
        super().__init__(name, power)
        self.weapon=weapon
    
    def weapon_details(self):
        print(f"I use {self.weapon}")   
```

```python
hammer=Thor("Thor-Thunder God","Strom & Thunder ","Mjölnir")
```

```python
# Multilevel Inheritence

class Grandparents:
    def __init__(self,nationality,baldness,anger_issues):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issues=anger_issues

    def self_grdprnts_info(self):
        print(f"Hello my nationality is {self.nationality} and my anger issue is that {self.anger_issues} ")

class Parents(Grandparents):
    def __init__(self,nationality,baldness,anger_issues,illness):
        super().__init__(nationality,baldness,anger_issues)
        self.illness=illness

    def illness_details(self):
        print(f"About my illness {self.illness}.")

class child(Parents):
    def __init__(self, nationality, baldness, anger_issues, illness,standard):
        super().__init__(nationality, baldness, anger_issues, illness)       
        self.standard=standard 

    def child_standard_details(self):
        print(f"About children  study {self.standard}.")
Parent_1=Parents("Indian","yes","no","yes-Diabities,TB,Heart Surgery")    
Parent_1.self_grdprnts_info()
Parent_1.illness_details()

child_1=child("Indian","yes","no","yes-Diabities,TB,Heart Surgery","college")
child_1.child_standard_details()
```

```python
# Multiple Inheritance

class Antman:
    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.suit=suit
        self.weapon=weapon
        self.nationality=nationality

    def introduction_avg(self):
        print(f"My name is {self.name} and powers are : {self.power} ")

class DRStrange:
    def __init__(self,name,power,suit,weapon,nationality="American"):
            self.name=name
            self.power=power
            self.suit=suit
            self.weapon=weapon
            self.nationality=nationality
    def weapon_details(self):
        print(f"My name is {self.name} and he carries weapons : {self.weapon} ")
```

```python
class Hulk(Antman,DRStrange):
    def __init__(self, name, power, suit, weapon,nationality,anger_issue):
        super().__init__(name, power, suit, weapon, nationality)
        self.anger_issue=anger_issue

    def anger_description(self):
        print(f"About anger issues he living with it : {self.anger_issue} ")
```

```python
hulk_guy=Hulk("Hulk","Strength and injury recover","Mark 85 Ironman Suit","Repulsor Beams and shield","american","yes")
```

```python
hulk_guy.introduction_avg()
```
