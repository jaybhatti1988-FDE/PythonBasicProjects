```python
# Nested List
nested_list=[[2,9,1,16],[6,9,1,0,55],[67,34,12,0],[44,55,11,0,9]]
print(nested_list[2][:])
print(nested_list[0][2])
```

```python
# Nested Tuples
nested_tuples=((2,9,1,16),(6,9,1,0,55),(67,34,12,0),(44,55,11,0,9))
```

```python
nested_tuples[0][:]
```

```python
# Iterating over a tule inside a Tuple
for i in nested_tuples:
    for j in i:
        print(j,sep="")
    print(" ")
```
