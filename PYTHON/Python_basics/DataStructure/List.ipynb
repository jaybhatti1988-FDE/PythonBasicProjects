## List Methods

#### 1. append()--> insert the value at the end of the list.



#### 2. insert()--->insert the value at a specific index location inside our list.



#### 3. remove()--> remove the value from our list that we are going to pass inside it.



#### 4. pop()---> remove the value from the list at a specific index location.



#### 5. index()--> it provides the index of the value that we are going to pass inside it if that value is present inside that list.



#### 6. sort()--> it is going to help us out with sorting the values in a ascending manner.



#### 7. reverse()--> it can reverse the values present inside our list i.e, value at last index will be at first index now.




#### 8. count()--> it counts the number of occurence of a specific value that we are going to pass inside our list and it returns that count of occurence.




#### 9. clear()--> for removing all of the values that we are storing inside our list.

## List comprehension

### We are going to use Looping statements inside the list so that we can store the values inside the list in an iterable format and in list comprehension conditional statements are allowed as well

## Syntax for List comprehension is

### [expression for item in iterable if condition]

### expression --> what we want to print
### for item in iterable---> looping statement
### if condition---> conditional statement

```python
# print 1 to 100 numbers using for loop list comprehension
list_Numberic_1to100Items=[no for no in range(1,101)]
list_Numberic_1to100Items
```

```python
list_Squareroot=[items**2 for items in range(1,21)]
list_Squareroot
```

```python
list_EvenSquareroot=[items**2 for items in range(1,21) if items%2==0]
list_EvenSquareroot
```

```python
list_String_Items=["Amit","Nagar","Aansh","Vinod","Baani","Appy","Truck","Plane"]
list_String_Items
length_list_String_Items=[len(words) for words in list_String_Items]
length_list_String_Items
```
