# Nested Dictionary

#### Nested Dictionary will be used for keeping a dictionary inside another dictionar

```python
# Here, Store multiple Student Information like name,rollNo,Std,marks,grade

Student_Dictionary={"Student_1":{"Name":"Vikas","RollNo":21,"STD":"7th","Marks":"68%","Grade":"First"},
                    "Student_2":{"Name":"Radha","RollNo":51,"STD":"8th","Marks":"78%","Grade":"Distinction"},
                    "Student_3":{"Name":"Ashok","RollNo":11,"STD":"9th","Marks":"88%","Grade":"Distinction"},
                    "Student_4":{"Name":"Kosha","RollNo":10,"STD":"4th","Marks":"93%","Grade":"Distinction"},
                    "Student_3":{"Name":"Pooja","RollNo":61,"STD":"6th","Marks":"58%","Grade":"Second"}
                    }
Student_Dictionary
```

```python
for outer_key,outer_value in Student_Dictionary.items():
    print(f"Outer Key is : { outer_key } and Outer Value is : { outer_value } ")
    for key_1,value_1 in outer_value.items():
        print(f"Inner Key is : { key_1 } and Inner Value is : { value_1 }")
```

## Dictionary Comprehension

#### Dictionary comprehension helps us with using Looping Statement along with Conditional Statement inside our Dictionary

#### Syntax for Dictionary Comprehension:

#### {expression for item in iterable if condition}

```python
comprehnsive_dictionary={x**2 for x in range (11)}
comprehnsive_dictionary
```

```python
new_comprehnsive_dictionary={x:x**2 for x in range (11)}
new_comprehnsive_dictionary
```

```python
even_comprehnsive_dictionary={x:x**2 for x in range (21) if x%2==0}
even_comprehnsive_dictionary
```
