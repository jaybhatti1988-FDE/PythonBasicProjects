#### (1) Implement a simple while-loop driven menu that keeps asking the user to choose an action until they press q to quit. Menu options
###### 1 — Add order amount to a running list (use input())
###### 2 — Show all orders and totals after applying discounts
###### q — Quit

#### (2) Use continue to re-show the menu after invalid input and break to exit on q.Constraints: Keep the loop logic simple and use only lists and loops (no functions required).


```python
#### (1) Implement a simple while-loop driven menu that keeps asking the user to choose an action until they press q to quit. Menu options
###### 1 — Add order amount to a running list (use input())
###### 2 — Show all orders and totals after applying discounts
###### q — Quit

orders=[]

while True:

    print("\n----- Order Menu -----")
    print("1 - Add Order")
    print("2 - Show All Orders and Totals")
    print("q - Quit")

    choice=input(" Enter your choice:- ")

# option 1 Add Order
    if choice=="1":
        try:
           order_amount=int(input("Enter Order Amount :")) 

           if order_amount<0:
               print("Invalid Input..! Enter Positive value")
           else:  
               orders.append(order_amount)
               print("Orders added Successfully...!")
        except ValueError:
            print("Invalid Input! Please enter a numeric value.") 

# option 2 display orders 
    elif choice=="2":

        if len(orders)==0:
           print("Any Orders not Available") 
        else:
            total_orders_revenue=0
            print(f"\n{'Order Amount':<15}{'Discount%':<15}{'Final Amount':<15}")
            print("-" * 45)

            for orders_amount in orders:
                
                if orders_amount>=2000:
                    discount=15
            
                elif orders_amount>=1500 :
                    discount=10
            
                elif orders_amount>=1000:
                    discount=7
                else:
                    discount=0

                # calculate discount price
                discount_price = (orders_amount*discount)/100

                # calculate Final Amount price after discount
                final_amount = orders_amount - discount_price 

                # calculate Total Amount price after revenuae all orders
                total_orders_revenue +=final_amount 

                print(f"{orders_amount:<15}"f"{str(discount) + '%':<15}"f"{final_amount:<15.2f}")

            # total calculation & formating
            print("-"*50)
            print(f"Total Orders Revenue Discounts: {total_orders_revenue:,.2f} ")

# option 3 Quit  
    elif choice.lower()=="q":
        print("Exit from Application")
        break
    else:
        print("Invalid Choice ...! Please Choose Option either 1 or 2 or q")

```
