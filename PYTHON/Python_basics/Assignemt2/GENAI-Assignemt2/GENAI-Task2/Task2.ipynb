# Task 2: Process Multiple Orders (for loop)

#### Given a list of order amounts:orders = [1200, 2500, 800, 1750, 3000] Use a for loop to apply the discount rules from Task 1 to each order and print a summary table showing: order_amount → discount% → final_amount Also compute and print the total revenue after discounts.
#### Extra (optional): Print the number of orders that received a discount (discount > 0).

```python
## Given a list of order amounts:orders = [1200, 2500, 800, 1750, 3000] Use a for 
# loop to apply the discount rules from Task 1 to each order and print a summary 
# table showing: order_amount → discount% → final_amount Also compute and print the 
# total revenue after discounts.
#  Extra (optional): Print the number of orders that received a discount (discount > 0).

orders=[1200,2500,800,1750,3000] # define orders

total_orders_revenue=0 # count total amount of orders
discount_order_count = 0 # count which orders gets discount 

print("Order Amount \tDiscount% \tFinal Amount") # formating summary
print("-"*50)

# calculate discount based on order amount
for order_amount in orders:
    if order_amount>=2000:
        discount=15
            
    elif order_amount>=1500 :
        discount=10
            
    elif order_amount>=1000:
        discount=7
           
    else:
        discount=0

# calculate discount price
    discount_price = (order_amount*discount)/100

# calculate Final Amount price after discount
    final_amount = order_amount - discount_price 

# calculate Total Amount price after revenuae all orders
    total_orders_revenue +=final_amount 

# check how many orders items discount value above 0%
    if discount >0:
        discount_order_count+=1

    print(f"{order_amount:<15}{str(discount) + '%':<15}{final_amount:<15.2f}")

# total calculation & formating
print("-"*50)
print(f"Total Orders Revenue Discounts: {total_orders_revenue:,.2f} ")
print("Number of Orders Received Discount:", discount_order_count)

            
```
