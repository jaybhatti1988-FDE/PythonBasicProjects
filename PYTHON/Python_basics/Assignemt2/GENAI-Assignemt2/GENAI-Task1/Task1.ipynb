```python
##### Write a program that reads an integer order_amount from the user using input().
##### Apply the following discount rules and print the final amount:

##### order_amount >= 2000 → 15% discount
##### 1500 <= order_amount < 2000 → 10% discount
##### 1000 <= order_amount < 1500 → 7% discount
##### Otherwise → 0% discount

##### Ensure you convert input to a number and handle the case when the input is non-numeric by displaying an error message and exiting the program.


# for handling invalid input errors 
try:

    order_input=int(input("Enter Order Amount"))


# for getting  discount based on order amount
    if order_input>=2000:
        discount_price=(order_input*15)/100
        final_amount=order_input-discount_price
        print("you get 15% discount after discount you need to pay : ",final_amount)

    elif order_input>=1500 :
        discount_price=(order_input*10)/100
        final_amount=order_input-discount_price
        print("you get 10% discount after discount you need to pay : ",final_amount)

    elif order_input>=1000:
        discount_price=(order_input*7)/100
        final_amount=order_input-discount_price
        print("you get 7% discount after discount you need to pay : ",final_amount)

    elif order_input>=0:
        print("you do not get any discount you have to full payment : ",order_input) 

    else:
        print("Invalid Input...! Please Try Again...!")   # handle negative value

except ValueError:
    print("Invalid Input Please enter valid integer value")   # handle invalid input   
```
