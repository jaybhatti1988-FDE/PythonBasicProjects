```python
#### (1) Given a list of daily sales:Python daily = [200, 150, 0, 400, 50, -1, 300] Iterate through it with a for loop and:

#### If a day's value is -1, treat it as corrupted data and break the loop 
# (stop processing).
### If a day's value is 0, treat it as a day with no sales and continue 
# (skip adding to revenue). For valid positive sales, add to total_sales and print the 
# running total.


daily=[200, 150, 0, 400, 50, -1, 300]
total_sales=0

for sales in daily:
    if sales==-1:
        print("corrupted data and stop processing")
        break

    if sales==0:
        print("No sales for this day. skipping ")
        continue
# add valid sales
    total_sales +=sales
    print("Current Sales:",sales)
    print("Running Totals:",total_sales)

print("Final Total Sales:",total_sales)
```

```python
#### (2) Print the final total after the loop completes (or stops due to corrupted data)


daily=[200, 150, 0, 400, 50, -1, 300]
total_sales=0

for sales in daily:
    if sales==-1:
        print("corrupted data and stop processing")
        break

    if sales==0:
        print("No sales for this day. ")
        continue
# add valid sales
    total_sales +=sales
    print("Running Totals :",total_sales)

print("Final Total Sales :",total_sales)
```
