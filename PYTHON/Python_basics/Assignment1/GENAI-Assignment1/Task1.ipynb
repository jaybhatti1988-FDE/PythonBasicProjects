# Task 3: Product Pricing (Dictionaries)

#### 1) Create a dictionary price_dict where keys are product names and values are prices (integers or floats). Include at least 6 entries.

#### 2) Write small code blocks to:

#### Add a new product with price to price_dict.
#### Update the price of an existing product.
#### Remove a product by name (handle the case when the product does not exist).

#### 3) Print the average price of all products (use only dictionary operations and basic arithmetic).

#### Extra (optional): Print the product with both the maximum and minimum prices.

```python
## (1) Create a dictionary price_dict where keys are product names and values 
# are prices (integers or floats). Include at least 6 entries.

price_dict={
    'Colgate':250,
    'Sugar':450,
    'Peanuts Oil':1390.56,
    'Cow Clarified Butter':2300.79,
    'Milk':100,
    'Cheese':56.89,
    'Juice':89.21,
    'Snacks':85,
    'Tea-Pouch':350,
    'Shampoo':550.91
}

print(price_dict)
```

```python
## (2) Write small code blocks to:

## Add a new product with price to price_dict.
## Update the price of an existing product.
## Remove a product by name (handle the case when the product does not exist).

## Add a new product with price to price_dict.
price_dict["T-shirt"]=2500

## Update the price of an existing product.
price_dict["Milk"]=250
print(price_dict)

## Remove a product by name (handle the case when the product does not exist).

remove_product=input('Enter product name')

if remove_product in price_dict: # check product available in dictionary or not
    del price_dict[remove_product]
    print(f"\n{remove_product} Item remove Successfully in Existing Dictionary\n ")
else:
    print(f"\n{remove_product} Item not Existing in Dictionary \n")

print(price_dict)

```

```python
## (3) Print the average price of all products (use only dictionary operations and basic 
# arithmetic).

average_price_dict=round(sum(price_dict.values())/len(price_dict), 2)
print(f"Given dictionary Average Value is that: {average_price_dict}")
```

```python
## Extra (optional): Print the product with both the maximum and minimum prices.

# min value

min_price_dict=min(price_dict,key=price_dict.get)
min_price_dict_values=min(price_dict.values())
print(f"Min Value: {min_price_dict}:{min_price_dict_values}")

# max value

max_price_dict=max(price_dict,key=price_dict.get)
max_price_dict_values=max(price_dict.values())
print(f"Max Value: {max_price_dict}:{max_price_dict_values}")
```

# Task 4: Combined Operations

#### 1) Using the products list and price_dict, create a list of tuples named catalog where each tuple is {product_name, price, category}
#### 2) From catalog, create a new dictionary category_to_products that maps each category to a list of product names in that category.
#### 3) Print all products that belong to the category that has the maximum number of products.

```python
## (1) Using the products list and price_dict, create a list of tuples named catalog
#  where each tuple is {product_name, price, category}

product_names_list = [
    'Colgate', 'Sugar', 'Peanuts Oil', 'Cow Clarified Butter',
    'Milk', 'Cheese', 'Juice', 'Snacks', 'Tea-Pouch', 'Shampoo'
]

price_dict = {
    'Colgate': 250,
    'Sugar': 450,
    'Peanuts Oil': 1390.56,
    'Cow Clarified Butter': 2300.79,
    'Milk': 100,
    'Cheese': 56.89,
    'Juice': 89.21,
    'Snacks': 85,
    'Tea-Pouch': 350,
    'Shampoo': 550.91
}

categories = [
    'Personal Care',
    'Grocery',
    'Cooking Oil',
    'Dairy',
    'Dairy',
    'Dairy',
    'Beverage',
    'Food',
    'Beverage',
    'Personal Care'
]

catelog=[]

for i in range(len(product_names_list)):
    product=product_names_list[i]
    price=price_dict[product]
    category=categories[i]
    catelog.append((product,price,category))

print(catelog)
```

```python
## (2) From catalog, create a new dictionary category_to_products that maps each 
# category to a list of product names in that category.

category_to_products={}

# map each categoryto list of product names

for product,price,category in catelog:
    if category not in category_to_products:
        category_to_products[category]=[]
    category_to_products[category].append(product)
print(category_to_products)
```

```python
# (3) Print all products that belong to the category that has the maximum number of products.

max_category=max(category_to_products,key=lambda category:len(category_to_products[category]))

print("category with maximum number of products : ",max_category)
print("Products :",category_to_products[max_category])
```
