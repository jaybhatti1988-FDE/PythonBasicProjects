# Filter()

##### Filter Function is going to retieve only those items whic are meeting a certain condition in an iterable
##### The filter function constructs an Iterator from elements of an iterable for which a condition remains True or for which a Function returns True Value

##### Syntax for Filter Function
##### list(filter(func, iterable))

```python
def even_function_value(natural_no):
    if natural_no%2==0:
        return True
```

```python
itrble_list=[1,5,9,2,12,65,76,21,50,54,33,97,20]
print(list(filter(even_function_value,itrble_list)))
```

```python
# Using lambda function
print(list(filter(lambda x:x%2==0,itrble_list)))
```

```python
print(list(filter(lambda x:x>7,itrble_list)))
```

```python
itr_list=[1,5,9,2,12,65,76,21,99,96,34,12,56,50,54,33,97,20]
print(list(filter(lambda x:x>7 and x%2 ==0,itr_list)))
```

```python
employees_list=[{'Name':'Atul','Age':30,'Qualification':'LLM'},
                {'Name':'Vinod','Age':32,'Qualification':'MBA'},
                {'Name':'Prem','Age':29,'Qualification':'Political Science'},
                {'Name':'Pooja','Age':23,'Qualification':'MCA'},
                {'Name':'Preeti','Age':25,'Qualification':'Fine Arts'},
                ]
```

```python
def age_gntr(employees_list):
    if employees_list['Age']>29:
        return True
```
