```python
## To create a function that converts temperature from celcius to Fahrenheit or[
# Fahrenheit to Celcius

## if temperature is in celcius , then celcius to fahrenheit conversion is:

def temperature_conversion(temp_value,temp_type):
    if temp_type=="C" or temp_type=="c":
        temparature=(temp_value*9/5)+32
        print(f"The Current Temparature is {temp_value}° Celcius to Convert in Fahrenheit" , str(temparature) +"°F")

    elif temp_type=="F" or temp_type=="f":
        temparature=(temp_value-32) *9/5
        print(f"The Current Temparature is {temp_value}° Fahrenheit to Convert in Celcius" , str(temparature) +"°C")   

    else:
        print("Invalid input")     
```

```python
Input_Temp_Value=int(input("Enter Temparature Value :- "))
Input_Temp_Type=input("Enter type of Temparature :- ")
temperature_conversion(Input_Temp_Value,Input_Temp_Type)
```

```python
# Define a function that behaves as a Password Strength Checker.

'''
A password is considered Strong if:

1. Its length is greater than 8 characters.
2. It contains at least one lowercase letter (a-z).
3. It contains at least one uppercase letter (A-Z).
4. It contains at least one digit (0-9).
5. It contains at least one special character (!,@,#,$,%,^,&,*,(,),_,{,},[,],)
'''
```

#### islower() ---> checks that whether if the current character is a lowercase character or not
#### isupper() ---> checks that whether if the current character is a upercase character or not
#### isdigit() ---> checks that whether if the current character is a digit or not

```python
def check_password_strength(password):
    if len(password)<8:
        print(f"Entered Current Password '{password}' is weak")

    elif not any (char.islower() for char in password):
         print(f"Entered Current Password '{password}' is weak") 

    elif not any (char.isupper() for char in password):
         print(f"Entered Current Password '{password}' is weak")   

    elif not any (char.isdigit() for char in password):
         print(f"Entered Current Password '{password}' is weak")    

    elif not any (char for char in '!, @, #, $, %, ^, &, *, (, ), _'):
         print(f"Entered Current Password '{password}' is weak")  

    else:
         print(f"Entered Current Password '{password}' is Strong")  


```

```python
Check_PWD=input("Enter password to check their strength")
check_password_strength(Check_PWD)
```

#### Palindrome String
###### Palindrome string is going to have characters similar to it from both the sides or even if you will reverse the string still you will be getting the given string itself

###### palindrome --> aba , aaabbaaa , zzyzz
###### Not a Palindrome --> Ab , abaa

```python
# Check Value Palindrome or not

def Palindrome_Value(Input_string):
    string_input_value=Input_string.lower().replace(" ","")

    if string_input_value==string_input_value[::-1]:
        print(f'The current value String "{string_input_value}" is Palindrome.')
    else:
        print(f'The current value String "{string_input_value}" is not Palindrome.')
    
```

```python
Input_palindrme_Value=input("Enter any String")
Palindrome_Value(Input_palindrme_Value)
```
