```python
def translator(text):
    """ Engish text translate into Hindi Language """

    if text=="Hello" or text=="hello":
        print(f"convert {text} into Hindi is Namaste")    
    
    if text=="Help" or text=="help":
        print(f"text convert {text} into Hindi is Madad")  

    if text=="Required" or text=="required":
        print(f"text convert {text} into Hindi is jaruri")         
```

##### For having the access to our Function and to provide the input to our function we can simpally call the function

##### Function calling is just like calling a waitor in a restaurant where in just like we call the waitor and provide the input as the food item and then waitor provide us with the food itself , like in a similar way we can call a function as well and can provide the input and can get the required output after calling the function only

```python
eng_input=input("Enter any words")
translator(eng_input)
```

```python
def translator_eng(eng_txt):
    
    if eng_txt=="Hello" or eng_txt=="hello":
        hindi_txt="Namaste"        

    elif eng_txt=="Help" or eng_txt=="help":
        hindi_txt="Madad"
    
    elif eng_txt=="Required" or eng_txt=="required":
        hindi_txt="jaruri"
    
    return hindi_txt       

```

```python
eng_words_input=input("Enter any words")
hindi_words=translator_eng(eng_words_input)
print(hindi_words)
```
