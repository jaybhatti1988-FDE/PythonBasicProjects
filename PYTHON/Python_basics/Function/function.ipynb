# Recursive Function

#### By a Recursive Function we can say that a Function is going to call itself again and again

```python
def FactorialNumber(factnum):
    if factnum==0 or factnum==1:
        FactorialNumber=1
    else:
        FactorialNumber=1
        for i in range(1,factnum+1):
            FactorialNumber*=i    
        print(f" The Given Number's {factnum} Factorial Number is that :-  {FactorialNumber} ")    
```

```python
factNumbers=input("Enter any value")

try:
    factNumbers=int(factNumbers)
    FactorialNumber(factNumbers)
except:
    print(f"Unnessary values {factNumbers} is not Allowed. ") 
```

### return num* factorial_recursive(num-1)
'''

##### base condition ---> which will stop the loop or which will let the function to stop calling itself in our case , base condition was ---> if i ==0 or i==1--> 1
##### num* factorial_recursive(num-1)
##### num==5
##### 5*factorial_recursive(4) & factorial_recursive(4)---> 4 *factorial_recursive(3)
##### factorial_recursive(3)---> 3*factorial_recursive(2)
##### factorial_recursive(2)---> 2*factorial_recursive(1)
'''

```python
def FactorialNumber(factnum):
    if factnum == 0 or factnum == 1:      
        return 1
    return factnum * FactorialNumber(factnum - 1)
```

```python
factNumbers = input("Enter any value: ")

try:
    factNumbers = int(factNumbers)
    if factNumbers < 0:
        print("Negative values are not Allowed....!")
    else:
        result = FactorialNumber(factNumbers)
        print(f"The Given Number's {factNumbers} Factorial Number is that :- {result}")
except ValueError:
    print(f"Unnecessary value '{factNumbers}' is not Allowed.")
```

### Function with Variable length Arguments

##### Positional Arguments--> Position arguments are being used so that we can pass multiple values inside our function as an argument irrespective of the number of values we have mentioned as the parameter inside our function

```python
def recursiveAdd(*num):
    add=0
    for i in num:
        add+=i
    print("The addition of numbers are",add)            
```

```python
recursiveAdd(56,78,67,55,89)
```

#### Keyword Arguments --> Takes the input as the keyword values and irrespective of any number of keyword values you are going to pass it wont matter

```python
def data_items(**user_data):
    for key,value in user_data.items():
        print(f"{key}: {value}")
```

```python
data_items(Name="Jay",age="35",gender="Male")
```

#### You guys want to pass keyword value pairs as your input along with positional arguments as well

```python
def data_position(*Position,**keywords):
    for i in Position:
        print(f"Positional Values are",i)
    for key,value in keywords.items():
        print(f" key is : { key } & value is : { value } ")
```

```python
data_position(1,8,9,0,5,name="Dipak",Age="41",Gender="Male")
```
