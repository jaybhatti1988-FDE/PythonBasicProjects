# Lambda Function

##### Lambda Functions are the One Liner Functions defined by the Lambda Keyword

##### Syntax of Lambda Function
##### lambda arguments:expression


##### lambda --> keyword which simpally let the python interpreter know that the current function is our Lambda Function
##### Arguments --> Input values for our Function
##### Expression ---> Logic ( Function Body )

##### They can have any number of Arguments but only one expression or only one Logic
##### They are used for Short Operations or as Arguments to Higher Order Functions

##### It can be treated as a Function without a Name and they Dont even require a Return Statement

```python
lambda_Function=lambda x:x**2
```
