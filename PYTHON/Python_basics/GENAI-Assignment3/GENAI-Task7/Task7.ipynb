```python
#### Create the following three small functions:

#### add_price(prices_list, price) → adds price to the list.
#### get_average_price(prices_list) → returns average price.
#### get_max_price(prices_list) → returns the maximum price.

#### Then create a simple menu:
#### 1 → Add price
#### 2 → Show average price
#### 3 → Show highest price
#### q → Quit

# adds price to the list

def add_price(price_list,price):
    price_list.append(price)

# Calculate Average Price
def get_avg_price(price_list):
    if len(price_list)==0:
        return 0

    return sum(price_list)/len(price_list)

# Calculate the maximum price.
def get_max_price(price_list):
    if len(price_list)==0:
        return 0
    return max(price_list)

# Empty list prices
prices_list = []

# MENU
while True:

    print("\n----- Price Menu -----")
    print("1 - Add Price")
    print("2 - Show Average Price")
    print("3 - Show Highest Price")
    print("q - Quit")

    choice=input("Enter your Choice")

# Add Price

    if choice=="1":

        try:
            price = float(input("Enter price: "))
            add_price(prices_list, price)
            print("Price added successfully!: ",price)
        except ValueError:
            print("Please enter a valid numeric price.")

# Show Average Price

    elif choice=="2":
        if len(prices_list) == 0:
            print("No prices available.")
        else:
            average_price = get_avg_price(prices_list)

            print("Average Price:", round(average_price, 2))

# Show Max Price

    elif choice=="3":
        if len(prices_list) == 0:
            print("No prices available.")
        else:
            highest_price = get_max_price(prices_list)
            print("Highest Price:", highest_price)
# Quit

    elif choice.lower() == "q":
        print("Program closed.")
        break
  
    # Invalid Choice
    else:
        print("Invalid choice. Please select 1, 2, 3, or q.")
```
