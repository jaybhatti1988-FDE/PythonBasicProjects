```python
## Write a function process_prices(prices) that:Takes a list of prices.Uses map() 
# + lambda to apply 10% discount to all prices.Uses filter() to keep only discounted 
# prices that are above 300.Returns both lists:discounted_prices filtered_prices Test 
# with: process_prices([100, 500, 900, 50, 750])

def process_prices(prices):

# apply 10% discount to all prices
    discounted_prices=list(map(lambda price:price-(price*10/100),prices))

# keep only discounted prices that are above 300
    filter_prices=list(filter(lambda price:price>300,discounted_prices))

# retun lists
    return discounted_prices,filter_prices

# Tests
discounted_prices,filter_prices=process_prices([100, 500, 900, 50, 750])

print("Discount Prices : ",discounted_prices)
print("Filter Prices : ",filter_prices)

```
