## Task 1 - Basic Function: Price After Discount

#### Returns the price after discount.
#### If discount_percent is missing, apply a default discount of 5%.
#### Test the function with:apply_discount(1000, 10) apply_discount(500)  # uses default discount
#### Extra (optional): Add a condition inside the function to ensure discount never exceeds 60%.

```python
## Returns the price after discount.If discount_percent is missing, apply a default
#  discount of 5%. Test the function with:apply_discount(1000, 10) apply_discount(500)
#  uses default discount

def discount_price(price,discount=5):
    discount_amount=(price*discount)/100
    final_price=price-discount_amount
    return final_price

# discount with 10%
print(discount_price(1000,10))

# discount with default 5%
print(discount_price(500))
```

```python
#### Extra (optional): Add a condition inside the function to ensure discount
#  never exceeds 60%.

def discount_price(price,discount=5):

    if discount>60:
        discount=60

    discount_amount=(price*discount)/100
    final_price=price-discount_amount
    return final_price

# discount with 10%
print(discount_price(1000,10))

# discount with default 5%
print(discount_price(500))

# discount with above 60%
print(discount_price(1000,75))
```
