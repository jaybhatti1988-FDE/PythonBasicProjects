#### Create a recursive function factorial(n) that:Returns the factorial of n.Handles two edge cases: n == 0 and n == 1.Prints an error message if n is negative Test with:
#### factorial(5),factorial(0),factorial(-3)

```python
#### Create a recursive function factorial(n) that:Returns the factorial of n.
# Handles two edge cases: n == 0 and n == 1.Prints an error message if n is 
# negative Test with:#### factorial(5),factorial(0),factorial(-3)

def factorial(n):

    # check negative numbers
    if n<0:
        print("Error: Factorial Numbers are not recognized for negative Numbers")
        return None
    
    # Base Case
    if n==0 or n==1:
        return 1
    
    # Recursive Case
    return n*factorial(n-1)

 # Operations
print(factorial(5))
print(factorial(0))
print(factorial(-3))
```
