```python
#### Create a lambda function gst that returns price after adding 18% GST.
#### gst = lambda price: price + (0.18 * price)
#### print(gst(100))  # should return 118
## Extra (optional): Create another lambda to compute final price after GST + discount
#  together.

#gst = lambda price: price + (0.18 * price)
gst=lambda price: price+(0.18*price)

print(f"Price After GST : ",gst(100))

# compute final price after GST + discount
final_price=lambda price,discount_percent:(price-(price*discount_percent/100))*1.18

print(f"Final Price After Discount with GST : ",final_price(1000,10))
```
