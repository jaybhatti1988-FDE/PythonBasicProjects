# Task 4 - Using map(): Apply GST to List of Prices

#### Given prices = [100, 250, 400, 1200, 50] Use map() with your gst lambda to generate a new list prices_with_gst.Print both lists: Original prices,Prices after GST

```python
#### Given prices = [100, 250, 400, 1200, 50] Use map() with your gst lambda to
#  generate a new list prices_with_gst.Print both lists: Original prices,Prices after 
# GST

# Original Prices

prices = [100, 250, 400, 1200, 50]

# Use Lambda function with gst lambda

gst=lambda price:price+(0.18*price)

# Apply GST using MAP function

price_with_gst=list(map(gst,prices))

# Print original and with GST 

print("Original Prices:", prices)
print("Prices after GST:", price_with_gst)

```
