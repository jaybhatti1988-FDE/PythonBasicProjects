## Task 5 - Create Product Info File (User Input)

##### Ask the user for 3 product names & their prices.
##### Write them into a new file products.txt in this format:
##### ProductName | Price
##### Read the file and print each line with proper formatting.

```python
##### Ask the user for 3 product names & their prices.
##### Write them into a new file products.txt in this format:
##### ProductName | Price
##### Read the file and print each line with proper formatting.

try:
    with open('product_data.txt','w') as file:
        for(i) in range(3):
            product_name=input(f"Enter Product {i + 1} Name : ")
            product_price=float(input(f"Enter {product_name} Price : "))

            file.write(f"{product_name}| {product_price:.2f}\n")

    print("\n Products added Successfully")

    # Read products.txt
    print("\n ------------ Product Information ------------")

    with open('product_data.txt','r') as file:

        for line in file:
            product_name,product_price=line.strip().split("|")
            print(f"Product Name:{product_name.strip()}| " f"Product Price:{product_price.strip()}")

except ValueError:
    print("Invalid Value...! Please enter Integer Value")
```
