```python
# You must not use exceptions beyond file-related safeguards here.Ask the user for a filename to open.
# If the file exists, read and print it.If it does not exist, print:
# “File not found. Please check the filename.”
# Use simple condition checks with os.path.exists() (allowed).

import os

file_name=input("Enter file nae to open : ")

if os.path.exists(file_name):
    file_check=open(file_name,"r")
    file_content=file_check.read()
    file_check.close()

    print("\n ------------ File Content ------------")
    print(file_content)
else:
    print("File not found. Verify file name Again")    
```
