# Task 4: Generate Summary Report from File

#### Using only file reading operations:

##### Read all sales values from sales_data.txt.
##### Convert them into integers.
##### Calculate and print:
##### Total Sales
##### Highest Sale
##### Lowest Sale
##### Average Sale

##### Do not use any advanced libraries.

```python
#### Using only file reading operations:
# Read all sales values from sales_data.txt.
# Convert them into integers.
# Calculate and print:
# Total Sales
# Highest Sale
# Lowest Sale
# Average Sale

sales=[]

try:
    with open("../sales_data.txt","r") as file:

        # read sales files
        for line in file:
            line=line.strip()
            if line:
                sales.append(int(line))

    # check sales data available or not

    if len(sales)>0:
        total_sales=sum(sales)
        highest_sales=max(sales)
        lowest_sales=min(sales)
        average_sales=total_sales/len(sales)

        print("------------Sales Summary Report-----------")
        print("Total Sales : ",total_sales)
        print("Highest Sales : ",highest_sales)
        print("Lowest Sales : ",lowest_sales)
        print("Average Sales : ",average_sales)
    else:
        print("Empty Sales Data")


except FileNotFoundError:
    print("File not found")

except ValueError:
    print("Invalid Value found in file")
```
