#### 1. Append these new sales to the same file:5000, 2500, 1700
#### (2.) After appending, reopen and print the entire updated file.
#### Extra (optional): Print the total number of lines after appending.

```python
#### (1.) Append these new sales to the same file:5000, 2500, 1700
#### (2.) After appending, reopen and print the entire updated file.

new_sales=[5000, 2500, 1700]

# Append file into existing file

with open ("../sales_data_2.txt","a") as file:
    for sale in new_sales:
        file.write(str(sale)+"\n")

# Reopen and Read update file

with open ("../sales_data_2.txt","r") as file:
    data=file.read()

    print("Updated Sales data :",data)

    
   
```
