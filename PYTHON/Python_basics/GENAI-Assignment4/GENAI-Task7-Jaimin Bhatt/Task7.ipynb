## Task 7: Mini Project - Export Discounted Prices

#### Create a dictionary:python prices = {  "Mouse": 500,    "Keyboard": 800,    "Monitor": 7000,     "Pendrive": 400,   "Camera": 5000 }

##### Ask the user for a discount percentage.Write discounted prices into discount_report.txt using:Product | Original Price | Discounted PriceAfter writing, read the file and print it to the terminal.Extra (optional): Write a summary at the bottom of the file:Total Items: X Average Discounted Price: Y

```python
# Create a dictionary:
# python
# prices = {
#     "Mouse": 500,
#     "Keyboard": 800
#     "Monitor": 7000,
#     "Pendrive": 400,
#     "Camera": 5000
# }

# Ask the user for a discount percentage.
# Write discounted prices into discount_report.txt using:
# Product | Original Price | Discounted Price
# After writing, read the file and print it to the terminal.
# Extra (optional): Write a summary at the bottom of the file:
# Total Items: X
# Average Discounted Price: Y

prices = {
   "Mouse": 500,
   "Keyboard": 800,
   "Monitor": 7000,
   "Pendrive": 400,
   "Camera": 5000
 }

 # Discount Percentage

discount_percent=float(input("Enter discount percentage (%) : "))

total_discount_price=0

# discount price mention in file
with open("discount_report.txt","w") as file:
    file.write(" Product | Original Price | Discount Price \n")
    file.write("--------------- Summary ---------------\n")

    for product,original_price in prices.items():
       

# Discount Calculate
        discount_amount= (original_price * discount_percent) / 100


# Discount Calculate price
        discount_price= original_price - discount_amount

# Add Total
        total_discount_price +=discount_price

# write in the file
        file.write(f"{product} | {original_price:.2f} | {discount_price:.2f}\n")

# calculate Average Price
    average_discount_price=total_discount_price /len(prices) 

# Summary

    file.write("\n--------------- Summary ---------------\n")
    file.write(f"Total Items: {len(prices)}\n")
    file.write(f"Average Discout Price : {average_discount_price:.2f}\n")

# Read File
with open("discount_report.txt","r") as file:
   file_content=file.read()

# Print file 
print("\n--------------- Discount Report ---------------\n")   
print(file_content)
```
