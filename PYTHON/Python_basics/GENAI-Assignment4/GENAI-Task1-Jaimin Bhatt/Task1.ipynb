## Task 1 - Write Sales Records to a File


#### Create a list of sales amounts:sales = [1200, 450, 980, 1500, 3000] Open a file named sales_data.txt in write mode.Write each sale on a new line in the file.Close the file, then reopen it and print its contents.Extra (optional): Write the data in comma-separated format instead of separate lines.

```python
## Create a list of sales amounts:sales = [1200, 450, 980, 1500, 3000] Open a file named sales_data.txt in
#  write mode.Write each sale on a new line in the file.Close the file, then reopen it and print its contents.

# list of sales amounts
sales=[1200, 450, 980, 1500, 3000]

# Open a file write mode
with open('../sales_data.txt','w') as file:
 

# write each sale in new line

    for sale in sales:
        file.write(str(sale)+"\n")

# Reopen file in read mode

with open('../sales_data.txt','r') as file:

# Read complete file
    data=file.read()

# Print file 
    print(data)
```

```python
## Extra (optional): Write the data in comma-separated format instead of separate lines.

# list of sales amounts
sales=[1200, 450, 980, 1500, 3000]

# Open a file write mode
with open('../sales_data_etc.txt','w') as file:
    file.write(",".join(map(str,sales)))

# Reopen file in read mode

with open('../sales_data_etc.txt','r') as file:

# Read complete file
    data=file.read()

# Print file 
    print(data)
```
