```python
## Using the same sales_data.txt:
## Read the entire file using .read() and print it.
## Read the first line using .readline().
## Read all lines using .readlines() and convert them into a list of integers. 
# Ensure proper formatting and cleanup of newline characters.

# Read file in using .read()

with open("../sales_data.txt","r") as file:

# Read complete file
    data=file.read()
    print("Entire File:")
    print(data)


# Read the first line usig .readline()
with open("../sales_data.txt","r") as file:
    first_line=file.readline().strip()
    print("first line : ",first_line)    

# Read all lines usig .readlines()
with open("../sales_data.txt","r") as file:
    file_lines=file.readlines()
    print("All line : ",file_lines) 

# convert all lines into list of integers
sales_list=[]

for line in file_lines:
    sales_list.append(int(line.strip()))
print("Sales List : ",sales_list)
```
