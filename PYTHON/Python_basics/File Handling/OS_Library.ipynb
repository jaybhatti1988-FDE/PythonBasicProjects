# OS Library

#### OS Library is able to interact with our Operating System and can let us work with Multiple Directories and Files

#### getcwd()--> It is capable of providing us with the idea about the current working directory
#### getcwd ----> get current working directory

```python
import os
```

```python
currentpath=os.getcwd()
print("Current Working Directory is :- ",currentpath )
```

#### mkdir() ---> Making new directory It helps to create new Directory

```python
new_path="New_Folder"
os.mkdir(new_path)
print(f"The new Directory {new_path} is created Successfully ")
```

```python
# listdir()---> Listing all the directions presents ---> List Directories
All_Itms=os.listdir('.')
print(All_Itms)
```

```python
## Join Multiple paths together

## os.path.join()---> It will join multiple files path together

dir_name="Folder_data"
file_name="File_Data"
Full_Path=os.path.join(dir_name,file_name)
print(f"The Full Path is that :- { Full_Path }")
```

```python
Current_Complete_Path=os.path.join(os.getcwd(),dir_name,file_name)
print(f"The complete path is that :- {Current_Complete_Path}")
```

```python
# OS.path.exists()

given_path="Users\bhatt\Documents\LLM\PYTHON\App.py"

if os.path.exists(given_path) :
    print("Path Exists")
else :   
    print("path not Exists")
```

##### os.path.isfile()---> This checks that whether if the current path is a file or not
##### os.path.isdir()---> This checks that whether if the current path is a directory or not

```python
# os.path.isfile()
# os.path.isdir()

file_os_Path='New_Folder'

if os.path.isfile(file_os_Path):
    print("The current file belongs to a file")

elif os.path.isdir(file_os_Path):
    print("The current file belongs to a Directory")   

else:
    print("Not found...! The current file neither belongs to file nor directory. ")     

```

##### The complete path<--> Absolute Path---> os.path.abspath()
##### os.path.abspath()--> It provides us with the complete path of the absolute path of the file or directory

```python
import os

new_file_name = 'Total_Words.txt'
complete_path = os.path.abspath(new_file_name)
print("The Complete absolute path is that :- ", complete_path)
```
