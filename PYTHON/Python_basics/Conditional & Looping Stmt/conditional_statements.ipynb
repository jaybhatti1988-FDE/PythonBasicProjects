# Conditional Statements

#### Conditional Statements works like Real Life scenario where if a certain condition is meeting then a statement of code or a block of code gets executed

```python
# driving licince for eligible years

a=int(input("Enter you Age"))
if a<=17:
    print(f"Sorry...!  you are {a} years old and you are not eligible for Driving licence.")
else:
    print(f"Yupiee...! you are {a} years old and you are eligible for Driving licence.")
```

```python
# Figure out who is the tallest among two friends.

keshav_height=int(input("Enter your heights in meter "))
vinay_height=int(input("Enter your heights in meter"))


if keshav_height>vinay_height:
    print("Keshav taller than vinay")
else:
    print("Vinay is taller than keshav")        
```

```python
# Figure out who is the tallest among three friends.

keshav_height=int(input("Enter your heights in meter "))
vinay_height=int(input("Enter your heights in meter"))
rishabh_height=int(input("Enter your heights in meter"))

if keshav_height>vinay_height>rishabh_height:
    print(f"Keshav height is {keshav_height}, Vinay height is {vinay_height} and Rishabh height is {rishabh_height} So, Keshav taller among all 3 friends.")
elif vinay_height>keshav_height>rishabh_height:
    print(f"Vinay height is {vinay_height} , Keshav height is {keshav_height}, Rishabh height is {rishabh_height} So, Vinay taller than among all 3 friends.")
else :
    print(f"Rishabh height is {rishabh_height} , Keshav height is {keshav_height} and Vinay height is {vinay_height} So, Rishabh taller than among all 3 friends.")           
```

#### Nested Conditional Statement

###### Nested Conditional Statements involve if statements inside another if statement

```python
# Check leap year or not

year=int(input("Enter any year :- "))

if year%4==0 or year%100==0 or year%400==0:
    print(f"Given {year} year is leap year. ")
else:
    print(f"Given {year} year is not leap year. ")    
```
