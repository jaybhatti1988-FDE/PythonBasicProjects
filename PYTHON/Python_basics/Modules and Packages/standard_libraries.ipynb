#### csv library---> Most of the time we are going to work with csv files so thats why knowledge with csv library functionality matters

```python
import csv
```

```python
with open('example.txt',mode='w',newline='') as write_file:
    writer=csv.writer(write_file)
    writer.writerow(['name','age'])
    writer.writerow(['Jay',35])
    writer.writerow(['Meet',31])
```

```python
with open('example.txt',mode='r') as read_file:
    reader=csv.reader(read_file)
    for row in reader:
        print(row)
```

## JSON library 

#### When we are going to work with Nosql Database data then the data inside the Nosql Database is going to be their in the Json Format and along with that when we are going to work with the datasets as well then again these dataset will be consisting of data in the json format itself

#### When we get the Json Data , we get in a Json string format so by using json library we can convert our string data into dictionary format for serializing our data

```python
import json
json_data={"Name":"Amar","Age":34}
STR_JSON=json.dumps(json_data)
print(STR_JSON)
print(type(STR_JSON))
```

```python
JSON_DICT=json.loads(STR_JSON)
print(JSON_DICT)
print(type(JSON_DICT))
```

## Regular Expression

#### Regular Expression is used to retrieve info based on a specific pattern or if the pattern matches

```python
import re
text_REG='Hello My Name is Jay and My PWD is AIM_909#'
Reg_pattern=r'\d+'
Reg_match=re.search(Reg_pattern,text_REG)
print(Reg_match.group())
```
