# Exception Handling

#### Exception Handling is a way in which if we are getting the Errors in programming or if the user entered input causes an error , then instead of Throwing an Error we can raise an exception which can let the user know that the input is Not Correct or the given input can come up as an Error.As we dont want to stop the execution of the Program , so we raise an error which will let the user know that the current input is Not correct but the Progra. execution wont stop .
#### Exception are events that disrupt the normal flow of a program . They occur when an error is encountered during program execution.

### try Except Block

#### try :                         {You are trying to run the operation which might throw an err}
####    result=a/b

#### except:                       {You are going to represent that if the err has been thrown that should happen}
####    print("The division is not valid")

#### Name Error

```python
b=25
a=b
c=b
try:
    a=b
except NameError as exception:
    print (exception)    
```

#### Zero Division Error

```python
den=input("Enter Denomination Value")

try:
    result=1/den
    print(result)
    a=c
except ZeroDivisionError as excepts:
    print(excepts)  

except NameError as name_err:
    print(name_err)  

except Exception as ex:
    print(ex)
```

#### If we are getting an error and we are not aware that which specific error type it is then we can use the primary exception class which is Exception

#### The Parent Exception type must be mentioned at the end or Last as it is the Parent class so only if the previous exception class will not catch the error then this final one will be able to catch it

#### try ---> try block is going to have the operation where we can expect an error to occur
#### except--> which is used to handle the error in a way such that the program exceution shouldnt stop
#### else--> It will be consisting of the code which we want to run if the try block operation is working without throwing any error
#### finally--> Finally  block gets executed once we are done with Everything and even if we are getting an error or Not getting an error still the code inside the finally block gets executed

#### if try----> code is throwing an Error---> Except---> code will be running

#### if try---> code is not throwing an Error---> Else---> code will be running

```python
num_value=int(input("Enter any value"))
try:
    result=1/num_value
except Exception as exvalue:
    print(exvalue) 
else:
    print("The division result value will be " , round(result,2))       
```
